## Question 12 — Electricity Bill Function

**Problem:**  
An electricity company charges different rates based on units consumed. Create a function that calculates the bill.

**Suggested rates (you may adjust slightly):**
- 0 – 100 units → Rs. 5 per unit
- 101 – 200 units → Rs. 10 per unit
- Above 200 units → Rs. 15 per unit

**Requirements:**
- Create a function that accepts the number of **units consumed**.
- Calculate and **return** the final electricity bill using conditions.
- Display the returned bill outside the function.

**Focus:** Function + arguments + conditions + `return`

````python
def electricity_bill(units):
    if units < 0:
        return 0
    if units <= 100:
        return units * 5
    elif units <= 200:
        return 100 * 5 + (units - 100) * 10
    return 100 * 5 + 100 * 10 + (units - 200) * 15

bill = electricity_bill(250)
print("Electricity Bill: Rs.", bill)
````

## Question 16 — Student Marks Processor

**Problem:**  
A teacher has a list of marks of all students in a class and wants key statistics.

**Requirements:**
- Create a function that accepts a **list of student marks**.
- Using a loop (or appropriate logic), calculate and **return**:
  - Highest mark
  - Lowest mark
  - Total marks
  - Average marks
- Display the results outside the function.

**Focus:** Function + list + loop + `return`

````python
def marks_processor(marks):
    if not marks:
        return 0, 0, 0, 0
    total = 0
    highest = marks[0]
    lowest = marks[0]
    for mark in marks:
        total += mark
        if mark > highest:
            highest = mark
        if mark < lowest:
            lowest = mark
    average = total / len(marks)
    return highest, lowest, total, average

highest, lowest, total, average = marks_processor([78, 85, 92, 67, 88])
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Total Marks:", total)
print("Average Marks:", average)
````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
def process_order(price, quantity, discount_rate, tax_rate):
    def calculate_subtotal():
        return price * quantity

    def calculate_discount(subtotal):
        return subtotal * discount_rate / 100

    def calculate_tax(amount):
        return amount * tax_rate / 100

    def calculate_final_amount():
        subtotal = calculate_subtotal()
        discounted_total = subtotal - calculate_discount(subtotal)
        return discounted_total + calculate_tax(discounted_total)

    return calculate_final_amount()

order_total = process_order(2000, 3, 10, 5)
print("Final Order Amount: Rs.", round(order_total, 2))
````

## Question 29 — Student Result Management System

**Problem:**  
Create a complete student result system using functions.

**The system should:**
1. Accept student information
2. Accept multiple subject marks
3. Calculate total
4. Calculate percentage
5. Determine grade
6. Determine pass/fail
7. Display a final result card

**Requirements:**
- Use **multiple functions**
- Use arguments and `return`
- Use lists
- Use conditions and loops
- Use at least one advanced concept (`*args`, `**kwargs`, nested functions, or lambda)
- Do **not** write the entire program inside one function

**Focus:** Combining multiple function concepts

````python
def get_student_info(**info):
    return info

def calculate_result(marks):
    total = sum(marks)
    percentage = total / (len(marks) * 100) * 100 if marks else 0
    return total, percentage

def get_grade(percentage):
    if percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    return "F"

def get_status(marks):
    return "Pass" if all(mark >= 40 for mark in marks) else "Fail"

def display_result(student, marks, total, percentage, grade, status):
    print("STUDENT RESULT CARD")
    for key, value in student.items():
        print(f"{key.title()}: {value}")
    print("Subject Marks:", marks)
    print("Total Marks:", total)
    print("Percentage:", round(percentage, 2), "%")
    print("Grade:", grade)
    print("Status:", status)

student = get_student_info(name="Osama Ali", roll_number="SMIT101", course="Python")
marks = [85, 78, 90, 82, 88]
total, percentage = calculate_result(marks)
grade = get_grade(percentage)
status = get_status(marks)
display_result(student, marks, total, percentage, grade, status)
````

````python
def customer_information(name, address):
    return {"Customer": name, "Address": address}

def product_information():
    return [
        {"name": "Laptop", "price": 120000, "quantity": 1},
        {"name": "Mouse", "price": 2500, "quantity": 2},
        {"name": "Keyboard", "price": 4500, "quantity": 1}
    ]

def calculate_item_total(price, quantity):
    return price * quantity

def calculate_subtotal(*items):
    return sum(items)

def calculate_discount(subtotal, rate=10):
    return subtotal * rate / 100

def calculate_tax(amount, rate=5):
    return amount * rate / 100

def calculate_delivery(subtotal):
    return 0 if subtotal >= 50000 else 500

def calculate_final_bill(subtotal, discount, tax, delivery):
    return subtotal - discount + tax + delivery

def create_order(**details):
    def format_amount(amount):
        return f"Rs. {amount:,.2f}"

    print("\n" + "=" * 48)
    print("                 ORDER SUMMARY")
    print("=" * 48)
    for key, value in details.items():
        print(f"{key}: {value}")
    print("=" * 48)

customer = customer_information("Osama Ali", "Karachi, Pakistan")
products = product_information()
products.sort(key=lambda item: item["price"], reverse=True)
item_totals = []

print("CUSTOMER:", customer["Customer"])
print("ADDRESS:", customer["Address"])
print("\nORDERED PRODUCTS")
print(f"{'Product':<15} {'Price':>10} {'Qty':>5} {'Total':>12}")

for product in products:
    item_total = calculate_item_total(product["price"], product["quantity"])
    item_totals.append(item_total)
    print(f"{product['name']:<15} {product['price']:>10,.2f} {product['quantity']:>5} {item_total:>12,.2f}")

subtotal = calculate_subtotal(*item_totals)
discount = calculate_discount(subtotal)
taxable_amount = subtotal - discount
tax = calculate_tax(taxable_amount)
delivery = calculate_delivery(subtotal)
final_amount = calculate_final_bill(subtotal, discount, tax, delivery)

create_order(
    Subtotal=f"Rs. {subtotal:,.2f}",
    Discount=f"Rs. {discount:,.2f}",
    Tax=f"Rs. {tax:,.2f}",
    Delivery=f"Rs. {delivery:,.2f}",
    Total=f"Rs. {final_amount:,.2f}",
    Payment="Cash on Delivery",
    Status="Order Confirmed"
)
````
